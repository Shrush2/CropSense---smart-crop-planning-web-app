```python
import pandas as pd
```

```python
price_data = pd.read_csv("dataset/crop_prices.csv")
price_data.head()
```

```python
def calculate_profit(crop, predicted_yield):

    row = price_data[price_data["crop"] == crop]

    price = row["price"].values[0]

    profit = predicted_yield * price

    return profit
```

```python
crop = input("Enter crop: ")
predicted_yield = float(input("Enter predicted yield: "))

profit = calculate_profit(crop, predicted_yield)

print("Estimated Profit:", profit)
```
